

### Plot 1: Average Flight Delay by Month

This plot shows the average arrival delay of flights for each month. It helps to identify which months have higher or lower flight delays.


```python
plt.figure(figsize=(8,5))

df.groupby("Month")["ArrDelay"].mean().plot(kind="bar",width=0.6 )

plt.title("Average Flight Delay by Month", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Average Arrival Delay (minutes)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

### Plot 2: Distance Distribution

This plot shows the distribution of flight distances. It helps to understand how the flight distances are spread across the dataset.


```python
plt.figure()
sns.histplot(df["Distance"].dropna(), bins=30, color="green")
plt.title("Distance Distribution")
plt.xlabel("Distance")
plt.show()
```

### Plot 3: Air Time Distribution

This plot shows the distribution of flight air time. It helps to understand how long flights usually remain in the air.


```python
plt.figure()
sns.histplot(df["AirTime"].dropna(), bins=30, color="orange")
plt.title("Air Time Distribution")
plt.xlabel("Air Time")
plt.show()
```

### Plot 4: Top 10 Origin Airports

This plot shows the top 10 airports based on the number of flights. It helps identify the airports with the highest number of departing flights.


```python
plt.figure(figsize=(10,5))
df["Origin"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Origin Airports")
plt.xlabel("Airport")
plt.ylabel("Flights")
plt.show()
```


### Plot 5: Top Airports with Highest Delay

This bar chart shows the top 10 origin airports with the highest average departure delay. It helps identify the airports where flights experience more delays on average.

### Plot 14: Year-wise Average Arrival Delay

This plot shows the average arrival delay for each year. It helps identify how flight delays have changed over the years.


```python
year_delay = df.dropna(subset=["ArrDelay"]).groupby("Year")["ArrDelay"].mean()

plt.figure(figsize=(10,5))
plt.plot(year_delay.index, year_delay.values, marker="o", color="red")
plt.title("Year-wise Average Arrival Delay")
plt.xlabel("Year")
plt.ylabel("Average Arrival Delay")
plt.grid(True)
plt.show()
```


### Plot 24: Cancelled vs Non-Cancelled Flights

This pie chart shows the percentage of cancelled and non-cancelled flights. It helps understand the proportion of flights that were cancelled.


### Plot 25: Delayed vs On-Time Flights

This pie chart shows the percentage of delayed and on-time flights. It helps understand the proportion of flights that arrived late compared to flights that arrived on time.

```python
status = df["ArrDelay"].apply(
    lambda x: "Delayed" if x > 0 else "On-Time"
)

status_count = status.value_counts()

plt.figure(figsize=(7,7))
plt.pie(
    status_count.values,
    labels=status_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Delayed vs On-Time Flights")
plt.show()
```
